# Understanding the Dynamics of and Simulating Starling Murmurations

## Introduction

In this repo, we seek to explore the dynamics of a starling murmuration. There are various models and understandings, ranging from naive to quite complex. There is no fixed scope to this project, rather, we seek to test and explore different models and parameters and observe the outcomes.

If you are unaware of the phenomenon of a [starling murmuration](https://www.youtube.com/watch?v=uV54oa0SyMc), it is a fascinating occurrence whose purpose is not definitively clear. It's widely believed that it is an anti predator measure.[1]

We have no evidence that any bird is a designated leader. Instead, it is believed that each bird responds to nearby birds and these relations cause the larger emergent patterns - hence we can call this a self-organising dynamical system. The main drivers of motion are separation, cohesion, and alignment; however, these in themselves cannot explain all of the observed behaviours. We will start our investigation here, and detail the mathematical functions following the remainder of the introduction.

As motivation, we will discuss some of the recent and historical findings towards understanding the behaviours within starling murmurations.

### Research History

#### STARFLAG [2]

An early breakthrough came in 2008 from the STARFLAG experiments in Rome, where researchers reconstructed the positions of individual birds in flocks of thousands of starlings. They found that flocks are typically thin, and not spherical, with reported dimensions roughly in the ratio $1:2.8:5.6$. They also found that flocks have sharply defined borders, maintain a minimum separation of approximately a wingspan, and that density is not uniform throughout the group.[2]

The most noteworthy result from this work, is that the interaction range is not based on all birds within a radius $r$. Instead, a bird responds to the movements of its nearest $k$ neighbours regardless of the desnity of the flock ($k$ is approximately 6 or 7). Hence, when we consider $\mathcal{N}_i(t)$, it will typically be defined as $$\mathcal{N}_i(t) = \{\text{the } k \text{ nearest birds to } i\},$$ and not $$\mathcal{N}_i(t) = \{j: \left\lVert x_j - x_i \right\rVert < R\}.$$ This notably impacts the dynamics in the presence of a predator, since neighbouring birds are not easily able to fall out of each others' interaction radii in this system.[3]

It has also been inferred that the nearest neighbours are distributed anisotropically (dependent on the bird's direction). That is to say, birds are less likely to have their nearest neighbour directly ahead or behind, than to the side. This anisotropy decays as more distant neighbours are considered.[3]

#### Local Actions to Global Correlations

Although each bird interacts only with its neighbours, measurements have shown that fluctuations in the motion of the murmuration are correlated over distances comparable with the entire size of the flock. That is to say, local interactions can support coordinated fluctuations across distances comparable with the size of the flock.[4]

Let $v_i$ be the velocity of bird $i$. Let its fluctuation relative to the flock mean be

$$\delta v_i = v_i - \bar{v}.$$

We can define a spatial correlation function, for example,

$$C(r) = \langle \delta v_i \cdot \delta v_j \rangle_{\left\lVert x_i - x_j\right\rVert \approx r},$$

which tells us how correlated the birds at a distance $r$ are on average. If this were plotted, we would see that for small $r$ values, the correlation is high, and it gradually tapers off as r grows. We call the distance over which the correlation persists the correlation length, $\xi$.

Cavagna and co [4] found that the correlation length $\xi$ scales approximately linearly with the flock's linear size $L$ - $\xi \propto L$. The authors interpreted this as the flocks operating near to a form of criticality, as seen in statistical-physics systems near phase transition, where small perturbations have long reaching effects. There exist maximum-entropy models fitted to starling data which have reproduced these scale-free correlations.[5]

#### Turning

In a starling flock, not all birds change direction simulatenously. A turn starts at a local point and travels through the flock. Detailed 3D tracking experiments were conducted [6], where researchers ranked birds according to when they began to turn. They found that directional information travelled (approximately) as $$x(t) = c_s t,$$ so the distance travelled by the turning signal was linear and not diffuse (i.e not like $x(t) \propto \sqrt{t}.$) The recorded propagation speed was roughly $c_s \approx 20 \text{ to } 40 \text{ m/s},$ while the birds were themselves flying at only around $ 7 \text{ to } 12 \text{ m/s}$. The researchers also noted that the turning "signal" demonstrated little attentuation as it travelled through the flock.[6]

The initial basic alignment model tends to behave like a damped relaxation process - a bird turns, the next turns partially and so on until the information "dies". Real flocks show behaviour much closer to that of a travelling wave, as seen in these results.[6]

#### Behavioural Inertia

Cavagna and co [6][7] believed that the missing input was turning inertia. In the standard model, direction is only adjusted by relaxing towards neighbouring directions, and no separate dynamical state recording the angular momentum and direction of the bird exists.



The inertial-spin model introduces such a concept, where each bird has

$$

x_i(t), \qquad v_i(t), \qquad s_i(t),

$$

where $s_i$ is an internal turning or "spin" variable.

The velocity direction then evolves as

$$\dot{v}_i = \frac{1}{\chi}s_i \times v_i,$$

while $s_i$ itself is a function of neighbouring velocities, damping, and noise.[7]

The overdamped limit (first-order) version of the model reduces towards our familiar alignment dynamics. With finite behavioural inertia it supports propagating turning waves.[7]

One attractive feature of this system is that the speed of any bird can remain approximately constant while its velocity vector rotates, matching the observation that collective turns are primarily changes in direction rather than accelerations and decelerations.[7]

#### Collective Turns

As you may be able to intuit from watching one, murmurations don't rotate as if the flock is a rigid object. Further research [8] reveals that neighbouring starlings tend to follow "equal-radius trajectories" - different birds may have different centres of rotation, but generally their turning radii are similar. Their trajectories can therefore cross, allowing birds to change their position relative to the overall geometry and direction of the flock, while maintaining much of the local neighbour structure.

Hence, a bird at a dangerous lateral tip before the turn can end up elsewhere after the turn. The same tracking work from earlier found that spontaneous turns more frequently originated from birds near the elongated tips of the flock. These particular birds showed larger and more persistent deviations from the mean direction before a turn began[8].

#### Predator Attacks

Predator attacks introduce a new layer of complexity onto the murmuration. Researchers distinguish between behaviours including: collective turns, flash expansions, agitation waves, vacuoles, blackening, compacting, dilution, cordons, and splits. It is important to note that different parts of one flock can perform different responses at any given time.[9][10]

Older field research analysing peregrine attacks indicated that the response depends somewhat on the behaviour of the given predator. For example, faster attacks were associated more strongly with flash expansions.[9]

A recent paper [10] from May 2026 tested behaviour under predatory influence by means of a remotely-controlled robotic falcon predator. They then took the data, and built a model where their simulated starlings interact with 7 topological neighbours using attraction, repulsion and alignment. They build on the model by giving individuals asynchronous behavioural states, such as normal flocking, alarmed flocking, escape, and a refractory period after escaping. Furthermore, proximity to the predator increases an internal stress variable; hence birds are able to initiate an escape themselves, or copy an escaping neighbour.

This model generates a varied collective behaviour, and argues that three features are especially important: how quickly the escape information spreads, the location of the birds relative tot he predator, and hysteresis (the current shape and density of the flock depend partly on what the flock was doing immediately beforehand).[10]

#### Other Interesting Notes and Recent Research

##### Dark waves

During an agitation wave, a dark band appears to move very quickly through the flock. It may understandably be thought that a region of increased bird density is travelling through the flock; however, simulations suggest that this is an "orientation wave". If birds bank or roll as they evade a predator, their projected wing area changes. If local birds copy such a manoeuvre, a wave of changes in orientation travels through the flock, providing such a phenomenon.[11] This is important, as it could entail the importance of modelling bird orientation, and not just location. A 2024 model called Flock2 [12] pushes this idea further by representing social influence as desired changes in orientation passed through an aerodynamic flight model rather than modelling Cartesian "social forces". This model is able to generate orientation waves and more realistic turning behaviour than our basic one.

An interesting aside, a recent paper was accepted by Physical Review Letters on the 3rd June 2026 regarding an improvement on the earlier inertial spin model.[13] The earlier inertial spin idea says the linear, low-attenuation propagation of collective turns resembles an underdamped spin wave.

These researchers' newer and higher resolution starling data posed a problem to this idea - they still observed underdamped travelling turning waves, while the ordinary unperturbed correlation function looked overdamped and Lorentzian, which shouldn't coexist in the simplest linear theory. Their proposed resolution is to add a nonlinear "Fermi-Pasta-Ulam-Tsingou-type term", so the travelling turning disturbance can behave more like a soliton than an ordinary linear spin wave. Such research is far beyond the scope of this project, and we now move to build the outlined initial model.[13]

## Building our first model

First we will build the basic model using "social forces" - separation, alignment, and cohesion. We will lay out the mathematical groundwork, and then explain the modelling specifics. Note here we will use $\mathcal{N}_i(t) = \{\text{the } k \text{ nearest birds to } i\}.$

### The Social Forces

#### Separation  
$$
\mathbf a_i^{\mathrm{sep}}
=
\sum_{j\in\mathcal N_i}
\frac{\mathbf x_i-\mathbf x_j}
{\|\mathbf x_i-\mathbf x_j\|^2}
$$

Accelerates the bird away from its local neighbours. Closer birds produce a stronger repulsive effect, with the magnitude of each contribution scaling inversely with distance.

#### Alignment 
$$
\mathbf a_i^{\mathrm{align}}
=
\frac{1}{k}
\sum_{j\in\mathcal N_i}
\mathbf v_j
-
\mathbf v_i
$$

Accelerates the bird towards the mean velocity of its neighbourhood, encouraging nearby birds to travel in similar directions and at similar speeds.

#### Cohesion
$$
\mathbf a_i^{\mathrm{coh}}
=
\frac{1}{k}
\sum_{j\in\mathcal N_i}
\mathbf x_j
-
\mathbf x_i
$$

Accelerates the bird towards the average position, or "middle", of its neighbourhood.

### Combining the Social Forces

The total acceleration of bird (i) is given by

$$
\mathbf a_i
=

\alpha \mathbf a_i^{\mathrm{sep}}
+
\beta \mathbf a_i^{\mathrm{align}}
+
\gamma \mathbf a_i^{\mathrm{coh}}.
$$

The three social-force terms have different physical dimensions, so the coefficients $\alpha$, $\beta$, and $\gamma$ are not dimensionless.

## Modelling

In order to build a model, we need to decide on the starting configuration, as well as some of the methodology.

We will model the entire configuration of birds as position and velocity matrices, with each row describing a single bird. We will initially use a flock of

$$
N = 1000 = 10^3.
$$


### Starting Position and Velocity

If all birds start equally spaced and at equal velocity, the simulation may remain overly regular. To avoid this, we add small random perturbations to both the birds' initial positions and velocities.

The initial positions will therefore be based on a regular grid, with a small random displacement added to each bird. Similarly, the birds will begin with approximately the same velocity, with a small random perturbation added to each velocity vector.

### Choosing the Neighbourhood

We must decide how each bird chooses its neighbourhood, as well as which value of (k) to use.

We choose

$$
k = 6,
$$

motivated by the estimates discussed above.

To begin with, we will naively compute the distance between every pair of birds at each time step. This has computational cost

$$
\mathcal{O}(N^2),
$$

after which we choose the 6 closest birds to each individual. For the flock sizes considered here, this should be acceptable.

### Scaling

We will use physical units throughout the simulation, with position measured in metres, velocity in metres per second, acceleration in metres per second squared, and time in seconds.

We take the initial spacing between grid points to be

$$
1 \text{ m}.
$$

This gives an initial flock approximately

$$
9 \text{ m}
$$

across in each direction before perturbations are added.

We take the initial mean velocity to be approximately

$$
10 \text{ m/s},
$$

broadly consistent with typical starling flight speeds.

## Building the Basic Model

```python
# Constants
N = 1000
K = 6
```

```python
import numpy as np

def calculate_acceleration(position, velocity, alpha, beta, gamma, K=6):
    """
    positions, velocitiy Nx3 vectors
    """

    # compute neighbourhoods
    differences = position[:, None, :] - position[None, :, :] # so differences[i,j] returns the vector of differences

    distances = np.linalg.norm(differences, axis=2) #NxN matrix of absolute diffs
    np.fill_diagonal(distances, np.inf) # exclude themselves

    neighbour_indices = np.argsort(distances, axis=1)[:, :K]


    neighbour_positions = position[neighbour_indices] # returns NxKx3 of all neighbour positions
    neighbour_velocities = velocity[neighbour_indices] # "      "            "         velocities

    alignment = np.mean(neighbour_velocities, axis=1) - velocity
    cohesion = np.mean(neighbour_positions, axis=1) - position


    bird_indices = np.arange(len(position))[:, None]

    neighbour_differences = differences[bird_indices, neighbour_indices]
    neighbour_distances = distances[bird_indices, neighbour_indices]

    separation = np.sum(
    neighbour_differences / neighbour_distances[:, :, None]**2,
    axis=1
    )

    acceleration = (
    alpha * separation
    + beta * alignment
    + gamma * cohesion
    )

    return acceleration
```

```python
def simulate_murmuration(
    T=10,
    dt=0.02,
    grid_size=6, #gridzize ^ 3 = N
    alpha=0.2,
    beta=1.0,
    gamma=1.0,
    K=6,
    seed=1):

    rng = np.random.default_rng(seed) #gen

    coordinates = np.arange(grid_size) #build initial grid
    grid = np.meshgrid(
    coordinates,
    coordinates,
    coordinates,
    indexing="ij") 

    grid = np.stack(grid, axis=-1)

    position = grid.reshape(-1, 3).astype(float) #randomise positions a little
    position += rng.uniform(
    -0.1,
    0.1,
    size=position.shape)   

    velocity = np.zeros_like(position) 
    velocity[:, 0] = 10 # initial velcoities

    velocity += rng.uniform(
    -0.5,
    0.5,
    size=velocity.shape)

    # run the loop 

    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    for step in range(n_steps):
        acceleration = calculate_acceleration(
        position,
        velocity,
        alpha,
        beta,
        gamma,
        K)

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```

```python
positions = simulate_murmuration(
    T=8,
    dt=0.03,
    grid_size=6,
    alpha=0.2,
    beta=1.0,
    gamma=1.0,
    K=6,
    seed=1
)

positions.shape
print(np.min(positions))
print(np.max(positions))
```

```python
# animation function 
def animate_murmuration(
    positions,
    dt,
    frame_skip=2,
    follow=False,
    save_path=None
): # ai generated

    if follow:
        frames = positions - np.mean(
            positions,
            axis=1,
            keepdims=True
        )
    else:
        frames = positions.copy()

    frames = frames[::frame_skip]

    # plot limits based on all frames
    mins = np.min(frames, axis=(0, 1))
    maxs = np.max(frames, axis=(0, 1))

    centre = (mins + maxs) / 2
    half_range = np.max(maxs - mins) / 2 * 1.1

    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection="3d")

    scatter = ax.scatter(
        frames[0, :, 0],
        frames[0, :, 1],
        frames[0, :, 2],
        s=6
    )

    ax.set_xlim(
        centre[0] - half_range,
        centre[0] + half_range
    )
    ax.set_ylim(
        centre[1] - half_range,
        centre[1] + half_range
    )
    ax.set_zlim(
        centre[2] - half_range,
        centre[2] + half_range
    )

    ax.set_box_aspect((1, 1, 1))

    def update(frame):

        current = frames[frame]

        scatter._offsets3d = (
            current[:, 0],
            current[:, 1],
            current[:, 2]
        )

        ax.set_title(
            f"t = {frame * frame_skip * dt:.2f} s"
        )

        return scatter,

    animation = FuncAnimation(
        fig,
        update,
        frames=len(frames),
        interval=50
    )

    # save animation if a path is given
    if save_path is not None:
        animation.save(
            save_path,
            writer="pillow",
            fps=20
        )

    plt.close(fig)

    return HTML(animation.to_jshtml())
```

## Adding Elliptical Starting Configuration and Speed Enforcement

Whilst these outputs are functional, they lack much of the distinct behaviour of a starling flock. As mentioned in the introduction, starling flocks tend not to be in a spherical or regular pattern, rather an elliptical shape which we can easily improve on. Furthermore, given our scaling, it's clear that our birds relax instantly to a speed far slower than that which they would actually move at. Hence we can add a speed enforcing term, which allows for some flucturation centered around the expected speed. 

We now redefine our functions accordingly.

```python
import numpy as np

def calculate_acceleration_speed_enforced(
    position,
    velocity,
    alpha,
    beta,
    gamma,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2
):
    """
    positions, velocity Nx3 vectors
    """

    # compute neighbourhoods
    differences = position[:, None, :] - position[None, :, :] # differences[i,j] returns vector of differences

    distances = np.linalg.norm(differences, axis=2) # NxN matrix of absolute diffs
    np.fill_diagonal(distances, np.inf) # exclude themselves

    neighbour_indices = np.argsort(distances, axis=1)[:, :K]

    neighbour_positions = position[neighbour_indices] # NxKx3 of all neighbour positions
    neighbour_velocities = velocity[neighbour_indices] # NxKx3 of all neighbour velocities

    alignment = np.mean(neighbour_velocities, axis=1) - velocity
    cohesion = np.mean(neighbour_positions, axis=1) - position

    bird_indices = np.arange(len(position))[:, None]

    neighbour_differences = differences[bird_indices, neighbour_indices]
    neighbour_distances = distances[bird_indices, neighbour_indices]

    separation = np.sum(
        neighbour_differences / (neighbour_distances[:, :, None]**2 + epsilon**2),
        axis=1
    )

    # PREFERRED SPEED
    speed = np.linalg.norm(velocity, axis=1, keepdims=True)

    speed_control = (
        speed_strength
        * (preferred_speed - speed)
        * velocity
        / np.maximum(speed, 1e-12)
    )

    acceleration = (
        alpha * separation
        + beta * alignment
        + gamma * cohesion
        + speed_control
    )

    return acceleration
```

```python
def simulate_murmuration_speed_enforced(
    T=10,
    dt=0.02,
    N=500,
    alpha=0.2,
    beta=1.0,
    gamma=1.0,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2,
    seed=1,
    min_spacing=0.5,
):

    rng = np.random.default_rng(seed)

    # ELLIPTICAL FLOCK randomly generated. checks if in the ellipsis, if not regenerate. 
    semi_axes = np.array([6.0, 3.0, 1.2])

    positions = []

    while len(positions) < N:

        candidate = rng.uniform(
            -semi_axes,
            semi_axes
        )

        # candidate must be inside ellipsoid
        inside = np.sum((candidate / semi_axes)**2) <= 1

        if not inside:
            continue

        # candidate must not be too close to an existing bird
        if len(positions) > 0:
            existing = np.array(positions)

            distances = np.linalg.norm(
                existing - candidate,
                axis=1
            )

            if np.min(distances) < min_spacing:
                continue

        positions.append(candidate)

    position = np.array(positions)

    # initial velocities
    velocity = np.zeros_like(position)
    velocity[:, 0] = preferred_speed

    velocity += rng.uniform(
        -0.5,
        0.5,
        size=velocity.shape
    )

    # run the loop

    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    for step in range(n_steps):

        acceleration = calculate_acceleration_speed_enforced(
            position,
            velocity,
            alpha,
            beta,
            gamma,
            K,
            preferred_speed,
            speed_strength
        )

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```

### Optimising Alignment Coefficient $\beta$

This is a clear improvement, but still moves linearly in one direction and doesn't have too much interesting behaviour. To further improve the model, we can try lowering the alignment, and maybe later try adding a feature which causes a turn in the outside birds. First we test different alignment coefficients.

```python
from IPython.display import display

betas = [1.0, 0.5, 0.3, 0.1]

positions_by_beta = {}

for beta in betas:

    print(f"Running beta = {beta}")

    positions = simulate_murmuration_speed_enforced(
        T=30,
        dt=0.02,
        N=300,
        alpha=0.2,
        beta=beta,
        gamma=1.0,
        K=6,
        preferred_speed=10.0,
        speed_strength=1.0,
        epsilon=0.2,
        seed=1,
        min_spacing=0.5
    )

    positions_by_beta[beta] = positions

    print(f"beta = {beta}")

    animation = animate_murmuration(
        positions,
        dt=0.02,
        frame_skip=5,
        follow=True
    )

    display(animation)
```

We see that they start to fragment and lose cohesion at low betas, but at high betas we get regular "gas like" behaviour. We can check if it ever settles to equilibrium via a metric:

$$
\bar{\mathbf{x}}(t)
=
\frac{1}{N}
\sum_{i=1}^{N}
\mathbf{x}_i(t)
$$

$$
R(t)
=
\sqrt{
\frac{1}{N}
\sum_{i=1}^{N}
\left\|
\mathbf{x}_i(t)-\bar{\mathbf{x}}(t)
\right\|^2
}
$$

```python
def flock_radius(positions):

    centres = np.mean(
        positions,
        axis=1,
        keepdims=True
    )

    centred = positions - centres

    radius = np.sqrt(
        np.mean(
            np.sum(centred**2, axis=2),
            axis=1
        )
    )

    return radius
```

```python
plt.figure(figsize=(8, 5))

for beta, positions in positions_by_beta.items():

    radius = flock_radius(positions)

    times = np.arange(len(radius)) * 0.02

    plt.plot(
        times,
        radius,
        label=f"beta = {beta}"
    )

plt.xlabel("Time (s)")
plt.ylabel("Flock radius (m)")
plt.legend()
plt.show()
```

## Adding Turning from Outside Behaviour

We see an interesting stability transition between 0.6 and 0.7, so for brevity we choose 0.65 as beta going forward. We can now attempt to simulate the turning of edge-situated birds in order to try and emulate the behaviour of the murmuration. We repatch our simulation functino to include such behaviour

We now introduce a simple local turning perturbation.

At a chosen time, we select a small proportion of birds from one elongated edge of the flock. These birds are then given an additional upward acceleration for a short period of time, after which the perturbation is removed.

This allows us to test whether a local change in direction can propagate through the flock using the existing interaction rules, or whether it is simply damped out or causes the flock to fragment.

```python
def simulate_murmuration_turn(
    T=30,
    dt=0.02,
    N=500,
    alpha=0.2,
    beta=0.65,
    gamma=1.0,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2,
    seed=1,
    min_spacing=0.5,
    turn_start=5.0,
    turn_duration=0.5,
    turn_strength=5.0,
    turn_fraction=0.1
):

    rng = np.random.default_rng(seed)

    # ELLIPTICAL FLOCK randomly generated
    semi_axes = np.array([6.0, 3.0, 1.2])

    positions = []

    while len(positions) < N:

        candidate = rng.uniform(
            -semi_axes,
            semi_axes
        )

        inside = np.sum((candidate / semi_axes)**2) <= 1

        if not inside:
            continue

        if len(positions) > 0:

            existing = np.array(positions)

            distances = np.linalg.norm(
                existing - candidate,
                axis=1
            )

            if np.min(distances) < min_spacing:
                continue

        positions.append(candidate)

    position = np.array(positions)

    # initial velocities
    velocity = np.zeros_like(position)
    velocity[:, 0] = preferred_speed

    velocity += rng.uniform(
        -0.5,
        0.5,
        size=velocity.shape
    )

    # run the loop
    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    turning_birds = None

    for step in range(n_steps):

        time = step * dt

        acceleration = calculate_acceleration(
            position,
            velocity,
            alpha,
            beta,
            gamma,
            K,
            preferred_speed,
            speed_strength,
            epsilon
        )

        # choose birds at one elongated tip when the turn begins
        if time >= turn_start and turning_birds is None:

            threshold = np.quantile(
                position[:, 0],
                1 - turn_fraction
            )

            turning_birds = np.where(
                position[:, 0] >= threshold
            )[0]

        # apply temporary upward acceleration
        if (
            turning_birds is not None
            and time < turn_start + turn_duration
        ):

            acceleration[turning_birds, 2] += turn_strength

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```

### Rebuilding the turns to be circular rather than an increase in velocity

We rebuild the turning mechanism so that selected edge birds are given a centripetal acceleration, causing their velocity direction to rotate while approximately preserving their speed.

#### Fixed-Radius Turning

Our previous turning mechanism applied a constant upward acceleration to a small group of birds. This caused some of the selected birds to separate from the flock rather than produce a realistic turn.

To model a turn more naturally, we instead use centripetal acceleration. For an object moving in a circle of radius $R$ with speed $v$, the required acceleration has magnitude

$$
a_\perp = \frac{v^2}{R}.
$$

This acceleration acts perpendicular to the current velocity and points towards the centre of rotation. As a result, the direction of the velocity changes while its magnitude remains approximately constant.

Equivalently, the angular velocity of the turn is

$$
\omega = \frac{v}{R}.
$$

For example, with a flight speed of approximately $10\text{ m/s}$ and a turning radius of $25\text{ m}$,

$$
\omega = \frac{10}{25} = 0.4\text{ rad/s}.
$$

If this turning acceleration is applied for $0.5$ seconds, the corresponding change in direction is approximately

$$
\theta = \omega t
= 0.4 \times 0.5
= 0.2\text{ rad}
\approx 11.5^\circ.
$$

We select a small group of birds from one lateral edge of the flock. For each selected bird, we define an individual centre of rotation lying approximately towards the centre of the flock and perpendicular to its current direction of travel. This allows the birds to follow similar-radius curved trajectories without requiring them all to rotate around exactly the same point.

```python
def simulate_murmuration_fixed_radius_turn(
    T=30,
    dt=0.02,
    N=500,
    alpha=0.2,
    beta=0.65,
    gamma=1.0,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2,
    seed=1,
    min_spacing=0.5,
    turn_start=5.0,
    turn_duration=0.5,
    turn_radius=25.0,
    turn_fraction=0.02
):

    rng = np.random.default_rng(seed)

    # ELLIPTICAL FLOCK randomly generated
    semi_axes = np.array([6.0, 3.0, 1.2])

    positions = []

    while len(positions) < N:

        candidate = rng.uniform(
            -semi_axes,
            semi_axes
        )

        inside = np.sum((candidate / semi_axes)**2) <= 1

        if not inside:
            continue

        if len(positions) > 0:

            existing = np.array(positions)

            distances = np.linalg.norm(
                existing - candidate,
                axis=1
            )

            if np.min(distances) < min_spacing:
                continue

        positions.append(candidate)

    position = np.array(positions)

    # initial velocities
    velocity = np.zeros_like(position)
    velocity[:, 0] = preferred_speed

    velocity += rng.uniform(
        -0.5,
        0.5,
        size=velocity.shape
    )

    # run the loop
    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    turning_birds = None
    turn_centres = None

    for step in range(n_steps):

        time = step * dt

        acceleration = calculate_acceleration(
            position,
            velocity,
            alpha,
            beta,
            gamma,
            K,
            preferred_speed,
            speed_strength,
            epsilon
        )

        # choose birds from one lateral edge when the turn begins
        if time >= turn_start and turning_birds is None:

            threshold = np.quantile(
                position[:, 1],
                1 - turn_fraction
            )

            turning_birds = np.where(
                position[:, 1] >= threshold
            )[0]

            # centre of the flock
            flock_centre = np.mean(
                position,
                axis=0
            )

            # direction from each turning bird towards the flock centre
            to_centre = (
                flock_centre
                - position[turning_birds]
            )

            # unit velocity vectors
            speeds = np.linalg.norm(
                velocity[turning_birds],
                axis=1,
                keepdims=True
            )

            velocity_direction = (
                velocity[turning_birds]
                / np.maximum(speeds, 1e-12)
            )

            # remove the component parallel to velocity
            # so the turning direction is perpendicular to motion
            perpendicular = (
                to_centre
                - np.sum(
                    to_centre * velocity_direction,
                    axis=1,
                    keepdims=True
                ) * velocity_direction
            )

            perpendicular_length = np.linalg.norm(
                perpendicular,
                axis=1,
                keepdims=True
            )

            turn_direction = (
                perpendicular
                / np.maximum(perpendicular_length, 1e-12)
            )

            # give each bird its own centre of rotation
            turn_centres = (
                position[turning_birds]
                + turn_radius * turn_direction
            )

        # apply centripetal acceleration during the turn
        if (
            turning_birds is not None
            and time < turn_start + turn_duration
        ):

            to_turn_centre = (
                turn_centres
                - position[turning_birds]
            )

            centre_direction = (
                to_turn_centre
                / np.maximum(
                    np.linalg.norm(
                        to_turn_centre,
                        axis=1,
                        keepdims=True
                    ),
                    1e-12
                )
            )

            speeds = np.linalg.norm(
                velocity[turning_birds],
                axis=1,
                keepdims=True
            )

            centripetal_acceleration = (
                speeds**2
                / turn_radius
            ) * centre_direction

            acceleration[turning_birds] += (
                centripetal_acceleration
            )

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```

```python
turn_positions = simulate_murmuration_fixed_radius_turn(
    T=30,
    dt=0.02,
    N=300,
    alpha=0.2,
    beta=0.65,
    gamma=1.0,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2,
    seed=1,
    min_spacing=0.5,
    turn_start=5.0,
    turn_duration=0.5,
    turn_radius=25.0,
    turn_fraction=0.02
)
```

## Basic Model Summary

Our first model was built from the three basic social forces of separation, alignment and cohesion:

$$
\mathbf a_i
=
\alpha \mathbf a_i^{\mathrm{sep}}
+
\beta \mathbf a_i^{\mathrm{align}}
+
\gamma \mathbf a_i^{\mathrm{coh}},
$$

with each bird interacting with its $K=6$ nearest neighbours. We later added a preferred speed term to prevent the flock relaxing to a low velocity, while the initial regular cube was replaced by a thin elliptical flock with a minimum initial separation. A softening parameter $\epsilon$ was also added to the separation term to avoid extremely large accelerations when two birds become very close.

The resulting model is capable of producing a coherent moving flock, as well as some non trivial internal rearrangement. Varying the alignment coefficient $\beta$ revealed a fairly clear change in stability. Lower values, such as $\beta=0.3$, allow local velocity differences to persist but eventually cause the flock to disperse or fragment. Higher values, such as $\beta=0.8$ or $\beta=1$, keep the flock bounded but rapidly damp much of the interesting internal motion. Using the flock radius

$$
R(t)
=
\sqrt{
\frac{1}{N}
\sum_{i=1}^{N}
\left\|
\mathbf{x}_i(t)-\bar{\mathbf{x}}(t)
\right\|^2
},
$$

we observed a change between these two regimes around $\beta=0.6$ to $0.7$, and chose $\beta=0.65$ as a compromise for the remaining experiments.

We then attempted to introduce collective turns. A simple upward acceleration applied to birds at one edge of the flock produced an unrealistic result: the affected birds could move far enough away to form their own neighbourhood and fragment from the main flock. We therefore replaced this with a fixed-radius turn using the centripetal acceleration

$$
a_\perp = \frac{v^2}{R},
$$

which changes the direction of travel while approximately preserving speed. This produces a more realistic local curved trajectory, but the disturbance still does not propagate through the flock in the wave-like manner observed in real starlings. Instead, the affected region tends to rearrange locally before the disturbance is damped out.

This appears to be the main limitation of the model rather than simply a poor choice of coefficients. Alignment acts as a first-order relaxation towards neighbouring velocities. Hence a directional disturbance is progressively averaged away rather than travelling through the flock with little attenuation. The model can therefore produce local deformation, stability and fragmentation, but the overall motion remains largely that of a coherent cloud travelling through space.

This motivates a move to a second model based on behavioural inertia. As discussed earlier, the inertial-spin model introduces an additional turning variable $\mathbf{s}_i$, so that changes in direction are themselves dynamical rather than instantaneous. In particular,

$$
\dot{\mathbf v}_i
=
\frac{1}{\chi}
\mathbf s_i \times \mathbf v_i,
$$

allows the velocity vector to rotate while approximately preserving its magnitude. The aim of Model 2 is therefore to investigate whether introducing this additional turning dynamics can produce the propagating, low-attenuation collective turns which Model 1 fails to capture.

## Model 2: The Inertial Spin Model

As discussed in the research history at the beginning of this notebook, observations of real starling flocks suggest that collective turns do not behave like simple diffusive relaxation. A turn can begin locally and propagate across the flock at a speed greater than the flight speed of the birds themselves, while exhibiting relatively little attenuation [6]. The first model does not reproduce this behaviour particularly well, since changes in direction are introduced directly through alignment and are therefore progressively averaged out.

To account for this, Cavagna et al. introduced the inertial-spin model, in which turning itself is treated as a dynamical process [7].

For each bird $i$, we now consider three state variables,

$$
\mathbf r_i(t), \qquad
\mathbf v_i(t), \qquad
\mathbf s_i(t),
$$

where

$$
\mathbf r_i(t) \in \mathbb R^3
$$

is position,

$$
\mathbf v_i(t) \in \mathbb R^3
$$

is velocity, and

$$
\mathbf s_i(t) \in \mathbb R^3
$$

is a generalized angular momentum associated with the turning of the velocity vector [7].

The variable $\mathbf s_i$ should not be interpreted as the ordinary angular momentum of the bird about some fixed point in space. Instead, it describes the instantaneous turning state of the bird. Its direction gives the axis about which the velocity vector is rotating, while its magnitude determines how rapidly that rotation occurs.

Ignoring stochastic noise initially, the deterministic dynamics are

$$
\dot{\mathbf r}_i
=
\mathbf v_i,
$$

$$
\dot{\mathbf v}_i
=
\frac{1}{\chi}
\mathbf s_i \times \mathbf v_i,
$$

and

$$
\dot{\mathbf s}_i
=
\frac{J}{v_0^2}
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}
\mathbf v_j
-
\frac{\eta}{\chi}
\mathbf s_i
$$

[7].

### Position dynamics

The first equation is simply

$$
\dot{\mathbf r}_i
=
\mathbf v_i.
$$

This is the usual definition of velocity. Once the velocity of bird $i$ is known, its position evolves according to this equation.

### Velocity and turning

The second equation is

$$
\dot{\mathbf v}_i
=
\frac{1}{\chi}
\mathbf s_i \times \mathbf v_i.
$$

The cross product

$$
\mathbf s_i \times \mathbf v_i
$$

is perpendicular to $\mathbf v_i$. Therefore, the acceleration generated by this term acts perpendicular to the current direction of motion.

This means that the equation changes the direction of the velocity vector without directly changing its magnitude.

To see this mathematically, consider

$$
\frac{d}{dt}\|\mathbf v_i\|^2.
$$

Since

$$
\|\mathbf v_i\|^2
=
\mathbf v_i\cdot\mathbf v_i,
$$

we obtain

$$
\frac{d}{dt}\|\mathbf v_i\|^2
=
2\mathbf v_i\cdot\dot{\mathbf v}_i.
$$

Substituting the velocity equation gives

$$
\frac{d}{dt}\|\mathbf v_i\|^2
=
\frac{2}{\chi}
\mathbf v_i\cdot
(\mathbf s_i\times\mathbf v_i).
$$

However,

$$
\mathbf s_i\times\mathbf v_i
\perp
\mathbf v_i,
$$

and therefore

$$
\mathbf v_i\cdot
(\mathbf s_i\times\mathbf v_i)
=
0.
$$

Hence

$$
\frac{d}{dt}\|\mathbf v_i\|^2
=
0,
$$

so

$$
\|\mathbf v_i(t)\|
=
v_0
$$

is constant under the idealised turning dynamics [7].

This is an important distinction from Model 1. We previously introduced an additional preferred-speed term in order to stop the flock from slowing down. In the inertial-spin formulation, constant speed follows directly from the geometry of the turning equation.

The parameter

$$
\chi
$$

is the generalized turning inertia. From

$$
\dot{\mathbf v}_i
=
\frac{1}{\chi}
\mathbf s_i\times\mathbf v_i,
$$

we can see that, for a fixed $\mathbf s_i$, a larger value of $\chi$ produces a slower change in direction, while a smaller value produces a faster change in direction.

If

$$
\mathbf s_i=0,
$$

then

$$
\dot{\mathbf v}_i=0,
$$

and the bird travels in a straight line.

If $\mathbf s_i$ is non-zero, the velocity vector rotates. For fixed speed $v_0$, the magnitude of the angular velocity is related to the spin by

$$
\omega_i
=
\frac{\|\mathbf s_i\|}{\chi}.
$$

Since circular motion satisfies

$$
v_0
=
R_i\omega_i,
$$

the corresponding turning radius is

$$
R_i
=
\frac{v_0\chi}{\|\mathbf s_i\|}.
$$

Thus a larger value of $\|\mathbf s_i\|$ corresponds to a tighter turn, while $\mathbf s_i=0$ corresponds to straight-line motion [7].

### Spin dynamics

The spin itself evolves according to

$$
\dot{\mathbf s}_i
=
\frac{J}{v_0^2}
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}
\mathbf v_j
-
\frac{\eta}{\chi}
\mathbf s_i.
$$

This equation contains two main terms.

The first is the interaction term,

$$
\frac{J}{v_0^2}
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}
\mathbf v_j.
$$

The quantity

$$
\sum_{j\in\mathcal N_i}\mathbf v_j
$$

represents the combined velocity of the birds in the neighbourhood of bird $i$.

If bird $i$ is aligned with its neighbours, then

$$
\mathbf v_i
\parallel
\sum_{j\in\mathcal N_i}\mathbf v_j,
$$

and therefore

$$
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}\mathbf v_j
=
0.
$$

In this case, no new spin is generated.

If bird $i$ is not aligned with its neighbours, then the cross product is non-zero. The magnitude of the cross product is

$$
\left\|
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}\mathbf v_j
\right\|
=
\|\mathbf v_i\|
\left\|
\sum_{j\in\mathcal N_i}\mathbf v_j
\right\|
\sin\theta_i,
$$

where $\theta_i$ is the angle between the bird's velocity and the combined neighbourhood velocity.

Hence directional disagreement generates a change in spin.

The parameter

$$
J
$$

controls the strength of this alignment interaction. For a fixed angular disagreement, a larger $J$ produces a larger value of $\dot{\mathbf s}_i$, and therefore causes the bird to develop a turning response more rapidly. A smaller $J$ produces a weaker response.

The factor

$$
\frac{1}{v_0^2}
$$

normalises the interaction by the characteristic flight speed. Since the cross product of two velocity vectors has magnitude proportional to velocity squared, this removes the trivial dependence of the interaction strength on the scale of $v_0$.

The second term is the damping term,

$$
-\frac{\eta}{\chi}\mathbf s_i.
$$

This acts in the opposite direction to the current spin and therefore reduces the turning state over time.

If the interaction term were zero, the spin equation would reduce to

$$
\dot{\mathbf s}_i
=
-\frac{\eta}{\chi}\mathbf s_i.
$$

This has solution

$$
\mathbf s_i(t)
=
\mathbf s_i(0)
e^{-(\eta/\chi)t}.
$$

Therefore spin decays exponentially, with characteristic timescale

$$
\tau
=
\frac{\chi}{\eta}.
$$

The parameter

$$
\eta
$$

therefore controls the damping of the turn. A larger $\eta$ causes spin to decay more rapidly, while a larger $\chi$ allows the turning state to persist for longer.

The three main parameters therefore have distinct roles:

$$
J:
\quad
\text{strength of alignment coupling},
$$

$$
\chi:
\quad
\text{turning inertia},
$$

$$
\eta:
\quad
\text{damping of the turning state}.
$$

### Reduction to heading dynamics

The reason this model can behave differently from Model 1 can be seen more clearly in two dimensions.

Suppose each bird travels at fixed speed $v_0$, so that its velocity can be written as

$$
\mathbf v_i
=
v_0
\begin{pmatrix}
\cos\theta_i \\
\sin\theta_i
\end{pmatrix},
$$

where $\theta_i$ is the heading angle.

In two dimensions, the spin has only one relevant component, perpendicular to the plane of motion. The relation between spin and angular velocity becomes

$$
s_i
=
\chi\dot{\theta}_i.
$$

Differentiating gives

$$
\dot{s}_i
=
\chi\ddot{\theta}_i.
$$

The interaction between two velocity directions depends on their angular difference. In particular,

$$
\mathbf v_i\times\mathbf v_j
\propto
\sin(\theta_j-\theta_i).
$$

For small differences in heading,

$$
\sin(\theta_j-\theta_i)
\approx
\theta_j-\theta_i.
$$

Substituting this small-angle approximation into the spin equation gives approximately

$$
\chi\ddot{\theta}_i
+
\eta\dot{\theta}_i
=
J
\sum_{j\in\mathcal N_i}
(\theta_j-\theta_i).
$$

This has the form of a system of coupled damped oscillators [7].

The important difference from Model 1 is the second-order term

$$
\chi\ddot{\theta}_i.
$$

A standard first-order alignment model has dynamics of the approximate form

$$
\dot{\theta}_i
=
c
\sum_{j\in\mathcal N_i}
(\theta_j-\theta_i),
$$

so differences in heading are directly relaxed towards the local mean. The inertial-spin model instead contains a second-order time derivative, meaning that the current rate of turning becomes part of the dynamics.

A bird which has developed non-zero spin does not immediately stop turning when the original directional disagreement begins to disappear. Its turning state persists until it is reduced by damping or modified by further interactions.

This additional inertia allows directional information to propagate through the flock in a wave-like manner rather than simply being smoothed out locally, which provides a possible explanation for the travelling collective turns discussed at the beginning of the notebook [6][7].

### Modelling changes

Our first implementation of the inertial-spin model used an initial condition in which every bird had exactly the same velocity,

$$
\mathbf v_i(0)
=
(v_0,0,0),
$$

and zero initial spin,

$$
\mathbf s_i(0)
=
\mathbf 0.
$$

This is an exact equilibrium of the directional dynamics. Since every bird is initially aligned with all of its neighbours,

$$
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}\mathbf v_j
=
\mathbf 0,
$$

and therefore

$$
\dot{\mathbf s}_i
=
\mathbf 0.
$$

Since the spin is also zero,

$$
\dot{\mathbf v}_i
=
\frac{1}{\chi}
\mathbf s_i\times\mathbf v_i
=
\mathbf 0.
$$

Before an external perturbation is applied, the flock therefore undergoes only rigid translation. In a reference frame following the centre of the flock, it appears completely stationary. This is mathematically consistent with the model, but it is not a particularly useful experiment for studying the propagation of directional information.

We therefore introduce a small amount of variation into the initial heading of each bird. The initial velocities remain concentrated around the positive $x$ direction, but are no longer perfectly parallel. Every velocity is then normalised so that

$$
\|\mathbf v_i(0)\| = v_0.
$$

The initial perturbations give the interaction term

$$
\frac{J}{v_0^2}
\mathbf v_i
\times
\sum_{j\in\mathcal N_i}\mathbf v_j
$$

a non-zero input, allowing the inertial-spin dynamics to act from the beginning of the simulation.

We retain a controlled turning perturbation later in the simulation. A small fraction of birds at one lateral edge of the flock are given an additional spin in the $z$ direction. Since the flock is initially travelling mainly in the positive $x$ direction, a $z$-directed spin rotates the corresponding velocity vectors in the $xy$ plane. We can then examine whether this local turning disturbance propagates through neighbouring birds, as observed experimentally in real collective turns [6].

At this stage we deliberately do not reintroduce the separation and cohesion terms from Model 1. The purpose of this experiment is first to isolate the inertial-spin mechanism and determine whether it transmits directional information through the flock. Adding additional spatial forces immediately would make it harder to distinguish this effect from the behaviour already produced by Model 1.

### Numerical and interaction corrections

Before changing the parameter regime, we make two corrections to the implementation of the inertial-spin model.

Firstly, selecting the $K$ nearest neighbours independently for every bird produces a directed interaction graph. In general, if bird $j$ is one of the nearest neighbours of bird $i$, it does not follow that bird $i$ is one of the nearest neighbours of bird $j$. The inertial-spin formulation, however, is based on a symmetric interaction between neighbouring birds.

Let $A$ denote the directed nearest-neighbour adjacency matrix, where

$$
A_{ij}
=
\begin{cases}
1, & j \in \mathcal N_i,\\
0, & \text{otherwise}.
\end{cases}
$$

We therefore replace this by the symmetric weighted interaction matrix

$$
W
=
\frac{1}{2}
\left(
A+A^\mathsf{T}
\right).
$$

A mutual neighbour relation has weight $1$, while a relation which exists in only one direction has weight $1/2$. This restores symmetry without increasing the mean interaction weight of the flock.

Secondly, our previous implementation updated the velocity using a forward Euler step and subsequently normalised it back to speed $v_0$. The continuous model instead rotates the velocity vector while preserving its magnitude exactly. We therefore update the spin first and then rotate the velocity directly using Rodrigues' rotation formula.

These changes do not introduce any new behavioural assumptions. They correct the interaction structure and numerical implementation before we investigate a different parameter regime.

```python
def rotate_velocity(velocity, spin, chi, dt):
    # angular velocity of each bird
    omega = spin / chi

    omega_size = np.linalg.norm(
        omega,
        axis=1,
        keepdims=True
    )

    # angle turned during this timestep
    angle = omega_size * dt

    # axis around which the velocity rotates
    axis = (
        omega
        / np.maximum(omega_size, 1e-12)
    )

    # rodrigues rotation formula
    rotated_velocity = (
        velocity * np.cos(angle)
        + np.cross(axis, velocity) * np.sin(angle)
        + axis
        * np.sum(axis * velocity, axis=1, keepdims=True)
        * (1 - np.cos(angle))
    )

    return rotated_velocity
```

```python
def simulate_inertial_spin(
    T=15,
    dt=0.005,
    N=300,
    K=6,
    v0=10.0,
    J=0.8,
    chi=1.25,
    eta=0.3,
    heading_noise=0.03,
    seed=1,
    min_spacing=0.5,
    turn_start=5.0,
    turn_fraction=0.02,
    spin_kick=0.15,
    turn_torque=0.0,
    turn_duration=0.0
):
    rng = np.random.default_rng(seed)

    # start with a thin ellipsoidal flock
    semi_axes = np.array([6.0, 3.0, 1.2])

    positions = []

    while len(positions) < N:
        candidate = rng.uniform(
            -semi_axes,
            semi_axes
        )

        # only keep points inside the ellipsoid
        if np.sum((candidate / semi_axes) ** 2) > 1:
            continue

        # stop birds starting too close together
        if len(positions) > 0:
            existing = np.array(positions)

            distances = np.linalg.norm(
                existing - candidate,
                axis=1
            )

            if np.min(distances) < min_spacing:
                continue

        positions.append(candidate)

    position = np.array(positions)

    # start mainly in the positive x direction
    direction = np.zeros_like(position)
    direction[:, 0] = 1.0

    # add small differences in initial heading
    direction[:, 1] += rng.normal(
        0,
        heading_noise,
        size=N
    )

    direction[:, 2] += rng.normal(
        0,
        heading_noise,
        size=N
    )

    # every bird starts at speed v0
    direction_size = np.linalg.norm(
        direction,
        axis=1,
        keepdims=True
    )

    velocity = (
        v0
        * direction
        / direction_size
    )

    # nobody is initially turning
    spin = np.zeros_like(position)

    n_steps = int(T / dt)

    positions_over_time = [position.copy()]
    velocities_over_time = [velocity.copy()]
    spins_over_time = [spin.copy()]

    turning_birds = None

    for step in range(n_steps):
        time = step * dt

        # calculate pairwise distances
        differences = (
            position[:, None, :]
            - position[None, :, :]
        )

        distances = np.linalg.norm(
            differences,
            axis=2
        )

        # stop each bird selecting itself
        np.fill_diagonal(
            distances,
            np.inf
        )

        # find the k nearest neighbours
        neighbour_indices = np.argsort(
            distances,
            axis=1
        )[:, :K]

        # build the directed neighbour matrix
        adjacency = np.zeros((N, N))

        adjacency[
            np.arange(N)[:, None],
            neighbour_indices
        ] = 1.0

        # use the symmetric part of the interaction
        weights = (
            adjacency
            + adjacency.T
        ) / 2

        neighbour_velocity_sum = (
            weights @ velocity
        )

        # select the birds which initiate the turn
        if time >= turn_start and turning_birds is None:
            threshold = np.quantile(
                position[:, 1],
                turn_fraction
            )

            turning_birds = np.where(
                position[:, 1] <= threshold
            )[0]

            # used for the small impulse experiment
            spin[turning_birds, 2] += spin_kick

        # neighbour disagreement generates spin
        interaction = (
            J / v0**2
        ) * np.cross(
            velocity,
            neighbour_velocity_sum
        )

        # existing spin is damped
        damping = (
            eta / chi
        ) * spin

        spin_change = (
            interaction - damping
        )

        # optionally apply a turning torque for a finite time
        if (
            turning_birds is not None
            and turn_start <= time < turn_start + turn_duration
        ):
            spin_change[
                turning_birds,
                2
            ] += turn_torque

        # update spin
        spin = (
            spin
            + spin_change * dt
        )

        # remove numerical spin parallel to velocity
        unit_velocity = velocity / v0

        spin = (
            spin
            - unit_velocity
            * np.sum(
                spin * unit_velocity,
                axis=1,
                keepdims=True
            )
        )

        # rotate velocity using the new spin
        velocity = rotate_velocity(
            velocity,
            spin,
            chi,
            dt
        )

        # update position
        position = (
            position
            + velocity * dt
        )

        positions_over_time.append(
            position.copy()
        )

        velocities_over_time.append(
            velocity.copy()
        )

        spins_over_time.append(
            spin.copy()
        )

    return (
        np.array(positions_over_time),
        np.array(velocities_over_time),
        np.array(spins_over_time),
        turning_birds
    )
```

```python
def measure_wave_speed(
    run,
    control,
    dt,
    turn_start,
    window,
    n_bands=12
):
    positions, velocities, spins, _ = run
    _, _, control_spins, _ = control

    turn_step = int(turn_start / dt)

    # remove the background dynamics
    spin_difference = (
        spins[:, :, 2]
        - control_spins[:, :, 2]
    )

    # distance from the perturbed edge when the turn begins
    y = positions[turn_step, :, 1]
    distance = y - np.min(y)

    # divide the flock into bands
    edges = np.quantile(
        distance,
        np.linspace(0, 1, n_bands + 1)
    )

    bands = np.digitize(
        distance,
        edges[1:-1]
    )

    band_distance = []
    peak_time = []

    end_step = min(
        len(spins),
        turn_step + int(window / dt)
    )

    for band in range(n_bands):
        birds = bands == band

        band_distance.append(
            np.mean(distance[birds])
        )

        signal = np.mean(
            spin_difference[
                turn_step:end_step,
                birds
            ],
            axis=1
        )

        peak = np.argmax(
            np.abs(signal)
        )

        peak_time.append(
            peak * dt
        )

    band_distance = np.array(band_distance)
    peak_time = np.array(peak_time)

    # measure distance relative to the first band
    band_distance -= band_distance[0]

    # leave out the source band itself
    distance = band_distance[1:]
    time = peak_time[1:]

    # fit d = c_s t through the origin
    speed = (
        np.sum(distance * time)
        / np.sum(time**2)
    )

    return speed
```

```python
J_values = np.array([
    0.8,
    8.0,
    32.0,
    128.0,
    209.0
])

measured_speeds = []

for J in J_values:
    print("running J =", J)

    # scale the measurement window with the expected wave timescale
    window = 12 * np.sqrt(0.8 / J)

    T = (
        5.0
        + window
        + 0.5
    )

    run = simulate_inertial_spin(
        T=T,
        dt=0.005,
        J=J,
        chi=1.25,
        eta=0.3,
        spin_kick=0.15,
        turn_torque=0.0,
        seed=1
    )

    control = simulate_inertial_spin(
        T=T,
        dt=0.005,
        J=J,
        chi=1.25,
        eta=0.3,
        spin_kick=0.0,
        turn_torque=0.0,
        seed=1
    )

    speed = measure_wave_speed(
        run,
        control,
        dt=0.005,
        turn_start=5.0,
        window=window
    )

    measured_speeds.append(speed)

measured_speeds = np.array(
    measured_speeds
)
```

```python
sqrt_J = np.sqrt(J_values)

# fit the predicted proportional relationship through the origin
gradient = (
    np.sum(
        sqrt_J * measured_speeds
    )
    / np.sum(sqrt_J**2)
)

x = np.linspace(
    0,
    np.max(sqrt_J) * 1.05,
    100
)

plt.figure(figsize=(6, 4))

plt.scatter(
    sqrt_J,
    measured_speeds
)

plt.plot(
    x,
    gradient * x
)

plt.xlabel(r"$\sqrt{J}$")
plt.ylabel(r"measured propagation speed $c_s$")
plt.title("turning-wave speed against alignment coupling")

plt.show()
```

### Model 2 Summary

The inertial-spin model resolves the main limitation identified in Model 1. Instead of directly relaxing each velocity towards those of its neighbours, directional disagreement first generates a turning state, which then rotates the velocity vector.

After correcting the interaction graph to use the symmetric weighting

$$
W
=
\frac{A+A^\mathsf{T}}{2},
$$

and replacing the Euler velocity update with a direct rotational update, a local spin perturbation was found to propagate through the flock rather than simply being damped locally. The response was approximately ballistic, in contrast to the diffusive behaviour associated with first-order alignment.

The linearised model predicts that the propagation speed satisfies

$$
c_s
\propto
\sqrt{\frac{J}{\chi}}.
$$

Holding $\chi$ fixed and varying $J$ produces the expected approximately linear relationship between $c_s$ and $\sqrt{J}$. The initial value $J=0.8$ therefore produced a genuine travelling disturbance, but on a timescale much slower than the motion of the flock. We consequently use a larger value, $J=209$, for the following simulations, giving a turning-wave speed of the same order as the flight speed $v_0=10$.

It is important that increasing $J$ changes the speed at which directional information propagates, but does not substantially increase the total angle through which the flock turns. The small instantaneous spin perturbation used above was deliberately useful as a diagnostic, but is not suitable for producing a visually significant collective turn.

We therefore now retain the validated inertial-spin dynamics and replace the small impulse with a finite-duration local turning torque.

## Producing a Visible Collective Turn

The previous section established that the inertial-spin dynamics can transmit a local change in direction across the flock. We now move away from diagnostic perturbations and construct a collective turn which is large enough to be visible in the spatial trajectories.

Rather than applying a single instantaneous spin kick, a small group of birds at one lateral edge is subjected to a turning torque for a finite interval. The torque enters through the spin equation, so the subsequent response of the remaining birds is still generated by the inertial-spin interaction rather than by directly forcing the whole flock.

We use the faster interaction regime identified above,

$$
J=209,\qquad
\chi=1.25,\qquad
\eta=0.3,
$$

with

$$
v_0=10.
$$

The aim of this section is no longer to diagnose the existence of the turning wave, but to examine the resulting collective trajectory and develop a visualisation which makes the three-dimensional motion of the flock apparent.
