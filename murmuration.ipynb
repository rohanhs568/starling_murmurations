# Understanding the Dynamics of and Simulating Starling Murmurations

## Introduction

In this repo, we seek to explore the dynamics of a starling murmuration. There are various models and understandings, ranging from naive to quite complex. There is no fixed scope to this project, rather, we seek to test and explore different models and parameters and observe the outcomes.

If you are unaware of the phenomenon of a [starling murmuration](https://www.youtube.com/watch?v=uV54oa0SyMc), it is a fascinating occurrence whose purpose is not definitively clear. It's widely believed that it is an anti predator measure.[1]

We have no evidence that any bird is a designated leader. Instead, it is believed that each bird responds to nearby birds and these relations cause the larger emergent patterns - hence we can call this a self-organising dynamical system. The main drivers of motion are separation, cohesion, and alignment; however, these in themselves cannot explain all of the observed behaviours. We will start our investigation here, and detail the mathematical functions following the remainder of the introduction.

As motivation, we will discuss some of the recent and historical findings towards understanding the behaviours within starling murmurations.

### Research History

#### STARFLAG [2]

An early breakthrough came in 2008 from the STARFLAG experiments in Rome, where researchers reconstructed the positions of individual birds in flocks of thousands of starlings. They found that flocks are typically thin, and not spherical, with reported dimensions roughly in the ratio $1:2.8:5.6$. They also found that flocks have sharply defined borders, maintain a minimum separation of approximately a wingspan, and that density is not uniform throughout the group.[2]

The most noteworthy result from this work, is that the interaction range is not based on all birds within a radius $r$. Instead, a bird responds to the movements of its nearest $k$ neighbours regardless of the desnity of the flock ($k$ is approximately 6 or 7). Hence, when we consider $\mathcal{N}_i(t)$, it will typically be defined as $$\mathcal{N}_i(t) = \{\text{the } k \text{ nearest birds to } i\},$$ and not $$\mathcal{N}_i(t) = \{j: \left\lVert x_j - x_i \right\rVert < R\}.$$ This notably impacts the dynamics in the presence of a predator, since neighbouring birds are not easily able to fall out of each others' interaction radii in this system.[3]

It has also been inferred that the nearest neighbours are distributed anisotropically (dependent on the bird's direction). That is to say, birds are less likely to have their nearest neighbour directly ahead or behind, than to the side. This anisotropy decays as more distant neighbours are considered.[3]

#### Local Actions to Global Correlations

Although each bird interacts only with its neighbours, measurements have shown that fluctuations in the motion of the murmuration are correlated over distances comparable with the entire size of the flock. That is to say, local interactions can support coordinated fluctuations across distances comparable with the size of the flock.[4]

Let $v_i$ be the velocity of bird $i$. Let its fluctuation relative to the flock mean be

$$\delta v_i = v_i - \bar{v}.$$

We can define a spatial correlation function, for example,

$$C(r) = \langle \delta v_i \cdot \delta v_j \rangle_{\left\lVert x_i - x_j\right\rVert \approx r},$$

which tells us how correlated the birds at a distance $r$ are on average. If this were plotted, we would see that for small $r$ values, the correlation is high, and it gradually tapers off as r grows. We call the distance over which the correlation persists the correlation length, $\xi$.

Cavagna and co [4] found that the correlation length $\xi$ scales approximately linearly with the flock's linear size $L$ - $\xi \propto L$. The authors interpreted this as the flocks operating near to a form of criticality, as seen in statistical-physics systems near phase transition, where small perturbations have long reaching effects. There exist maximum-entropy models fitted to starling data which have reproduced these scale-free correlations.[5]

#### Turning

In a starling flock, not all birds change direction simulatenously. A turn starts at a local point and travels through the flock. Detailed 3D tracking experiments were conducted [6], where researchers ranked birds according to when they began to turn. They found that directional information travelled (approximately) as $$x(t) = c_s t,$$ so the distance travelled by the turning signal was linear and not diffuse (i.e not like $x(t) \propto \sqrt{t}.$) The recorded propagation speed was roughly $c_s \approx 20 \text{ to } 40 \text{ m/s},$ while the birds were themselves flying at only around $ 7 \text{ to } 12 \text{ m/s}$. The researchers also noted that the turning "signal" demonstrated little attentuation as it travelled through the flock.[6]

The initial basic alignment model tends to behave like a damped relaxation process - a bird turns, the next turns partially and so on until the information "dies". Real flocks show behaviour much closer to that of a travelling wave, as seen in these results.[6]

#### Behavioural Inertia

Cavagna and co [6][7] believed that the missing input was turning inertia. In the standard model, direction is only adjusted by relaxing towards neighbouring directions, and no separate dynamical state recording the angular momentum and direction of the bird exists.



The inertial-spin model introduces such a concept, where each bird has

$$

x_i(t), \qquad v_i(t), \qquad s_i(t),

$$

where $s_i$ is an internal turning or "spin" variable.

The velocity direction then evolves as

$$\dot{v}_i = \frac{1}{\chi}s_i \times v_i,$$

while $s_i$ itself is a function of neighbouring velocities, damping, and noise.[7]

The overdamped limit (first-order) version of the model reduces towards our familiar alignment dynamics. With finite behavioural inertia it supports propagating turning waves.[7]

One attractive feature of this system is that the speed of any bird can remain approximately constant while its velocity vector rotates, matching the observation that collective turns are primarily changes in direction rather than accelerations and decelerations.[7]

#### Collective Turns

As you may be able to intuit from watching one, murmurations don't rotate as if the flock is a rigid object. Further research [8] reveals that neighbouring starlings tend to follow "equal-radius trajectories" - different birds may have different centres of rotation, but generally their turning radii are similar. Their trajectories can therefore cross, allowing birds to change their position relative to the overall geometry and direction of the flock, while maintaining much of the local neighbour structure.

Hence, a bird at a dangerous lateral tip before the turn can end up elsewhere after the turn. The same tracking work from earlier found that spontaneous turns more frequently originated from birds near the elongated tips of the flock. These particular birds showed larger and more persistent deviations from the mean direction before a turn began[8].

#### Predator Attacks

Predator attacks introduce a new layer of complexity onto the murmuration. Researchers distinguish between behaviours including: collective turns, flash expansions, agitation waves, vacuoles, blackening, compacting, dilution, cordons, and splits. It is important to note that different parts of one flock can perform different responses at any given time.[9][10]

Older field research analysing peregrine attacks indicated that the response depends somewhat on the behaviour of the given predator. For example, faster attacks were associated more strongly with flash expansions.[9]

A recent paper [10] from May 2026 tested behaviour under predatory influence by means of a remotely-controlled robotic falcon predator. They then took the data, and built a model where their simulated starlings interact with 7 topological neighbours using attraction, repulsion and alignment. They build on the model by giving individuals asynchronous behavioural states, such as normal flocking, alarmed flocking, escape, and a refractory period after escaping. Furthermore, proximity to the predator increases an internal stress variable; hence birds are able to initiate an escape themselves, or copy an escaping neighbour.

This model generates a varied collective behaviour, and argues that three features are especially important: how quickly the escape information spreads, the location of the birds relative tot he predator, and hysteresis (the current shape and density of the flock depend partly on what the flock was doing immediately beforehand).[10]

#### Other Interesting Notes and Recent Research

##### Dark waves

During an agitation wave, a dark band appears to move very quickly through the flock. It may understandably be thought that a region of increased bird density is travelling through the flock; however, simulations suggest that this is an "orientation wave". If birds bank or roll as they evade a predator, their projected wing area changes. If local birds copy such a manoeuvre, a wave of changes in orientation travels through the flock, providing such a phenomenon.[11] This is important, as it could entail the importance of modelling bird orientation, and not just location. A 2024 model called Flock2 [12] pushes this idea further by representing social influence as desired changes in orientation passed through an aerodynamic flight model rather than modelling Cartesian "social forces". This model is able to generate orientation waves and more realistic turning behaviour than our basic one.

An interesting aside, a recent paper was accepted by Physical Review Letters on the 3rd June 2026 regarding an improvement on the earlier inertial spin model.[13] The earlier inertial spin idea says the linear, low-attenuation propagation of collective turns resembles an underdamped spin wave.

These researchers' newer and higher resolution starling data posed a problem to this idea - they still observed underdamped travelling turning waves, while the ordinary unperturbed correlation function looked overdamped and Lorentzian, which shouldn't coexist in the simplest linear theory. Their proposed resolution is to add a nonlinear "Fermi-Pasta-Ulam-Tsingou-type term", so the travelling turning disturbance can behave more like a soliton than an ordinary linear spin wave. Such research is far beyond the scope of this project, and we now move to build the outlined initial model.[13]

## Building our first model

First we will build the basic model using "social forces" - separation, alignment, and cohesion. We will lay out the mathematical groundwork, and then explain the modelling specifics. Note here we will use $\mathcal{N}_i(t) = \{\text{the } k \text{ nearest birds to } i\}.$

### The Social Forces

#### Separation  
$$
\mathbf a_i^{\mathrm{sep}}
=
\sum_{j\in\mathcal N_i}
\frac{\mathbf x_i-\mathbf x_j}
{\|\mathbf x_i-\mathbf x_j\|^2}
$$

Accelerates the bird away from its local neighbours. Closer birds produce a stronger repulsive effect, with the magnitude of each contribution scaling inversely with distance.

#### Alignment 
$$
\mathbf a_i^{\mathrm{align}}
=
\frac{1}{k}
\sum_{j\in\mathcal N_i}
\mathbf v_j
-
\mathbf v_i
$$

Accelerates the bird towards the mean velocity of its neighbourhood, encouraging nearby birds to travel in similar directions and at similar speeds.

#### Cohesion
$$
\mathbf a_i^{\mathrm{coh}}
=
\frac{1}{k}
\sum_{j\in\mathcal N_i}
\mathbf x_j
-
\mathbf x_i
$$

Accelerates the bird towards the average position, or "middle", of its neighbourhood.

### Combining the Social Forces

The total acceleration of bird (i) is given by

$$
\mathbf a_i
=

\alpha \mathbf a_i^{\mathrm{sep}}
+
\beta \mathbf a_i^{\mathrm{align}}
+
\gamma \mathbf a_i^{\mathrm{coh}}.
$$

The three social-force terms have different physical dimensions, so the coefficients $\alpha$, $\beta$, and $\gamma$ are not dimensionless.

## Modelling

In order to build a model, we need to decide on the starting configuration, as well as some of the methodology.

We will model the entire configuration of birds as position and velocity matrices, with each row describing a single bird. We will initially use a flock of

$$
N = 1000 = 10^3.
$$


### Starting Position and Velocity

If all birds start equally spaced and at equal velocity, the simulation may remain overly regular. To avoid this, we add small random perturbations to both the birds' initial positions and velocities.

The initial positions will therefore be based on a regular grid, with a small random displacement added to each bird. Similarly, the birds will begin with approximately the same velocity, with a small random perturbation added to each velocity vector.

### Choosing the Neighbourhood

We must decide how each bird chooses its neighbourhood, as well as which value of (k) to use.

We choose

$$
k = 6,
$$

motivated by the estimates discussed above.

To begin with, we will naively compute the distance between every pair of birds at each time step. This has computational cost

$$
\mathcal{O}(N^2),
$$

after which we choose the 6 closest birds to each individual. For the flock sizes considered here, this should be acceptable.

### Scaling

We will use physical units throughout the simulation, with position measured in metres, velocity in metres per second, acceleration in metres per second squared, and time in seconds.

We take the initial spacing between grid points to be

$$
1 \text{ m}.
$$

This gives an initial flock approximately

$$
9 \text{ m}
$$

across in each direction before perturbations are added.

We take the initial mean velocity to be approximately

$$
10 \text{ m/s},
$$

broadly consistent with typical starling flight speeds.

```python
# Constants
N = 1000
K = 6
```

```python
import numpy as np

def calculate_acceleration(position, velocity, alpha, beta, gamma, K=6):
    """
    positions, velocitiy Nx3 vectors
    """

    # compute neighbourhoods
    differences = position[:, None, :] - position[None, :, :] # so differences[i,j] returns the vector of differences

    distances = np.linalg.norm(differences, axis=2) #NxN matrix of absolute diffs
    np.fill_diagonal(distances, np.inf) # exclude themselves

    neighbour_indices = np.argsort(distances, axis=1)[:, :K]


    neighbour_positions = position[neighbour_indices] # returns NxKx3 of all neighbour positions
    neighbour_velocities = velocity[neighbour_indices] # "      "            "         velocities

    alignment = np.mean(neighbour_velocities, axis=1) - velocity
    cohesion = np.mean(neighbour_positions, axis=1) - position


    bird_indices = np.arange(len(position))[:, None]

    neighbour_differences = differences[bird_indices, neighbour_indices]
    neighbour_distances = distances[bird_indices, neighbour_indices]

    separation = np.sum(
    neighbour_differences / neighbour_distances[:, :, None]**2,
    axis=1
    )

    acceleration = (
    alpha * separation
    + beta * alignment
    + gamma * cohesion
    )

    return acceleration
```

```python
def simulate_murmuration(
    T=10,
    dt=0.02,
    grid_size=6, #gridzize ^ 3 = N
    alpha=0.2,
    beta=1.0,
    gamma=1.0,
    K=6,
    seed=1):

    rng = np.random.default_rng(seed) #gen

    coordinates = np.arange(grid_size) #build initial grid
    grid = np.meshgrid(
    coordinates,
    coordinates,
    coordinates,
    indexing="ij") 

    grid = np.stack(grid, axis=-1)

    position = grid.reshape(-1, 3).astype(float) #randomise positions a little
    position += rng.uniform(
    -0.1,
    0.1,
    size=position.shape)   

    velocity = np.zeros_like(position) 
    velocity[:, 0] = 10 # initial velcoities

    velocity += rng.uniform(
    -0.5,
    0.5,
    size=velocity.shape)

    # run the loop 

    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    for step in range(n_steps):
        acceleration = calculate_acceleration(
        position,
        velocity,
        alpha,
        beta,
        gamma,
        K)

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```

```python
def animate_murmuration(
    positions,
    dt,
    frame_skip=2,
    follow=False,
    save_path=None
): # ai generated

    if follow:
        frames = positions - np.mean(
            positions,
            axis=1,
            keepdims=True
        )
    else:
        frames = positions.copy()

    frames = frames[::frame_skip]

    # plot limits based on all frames
    mins = np.min(frames, axis=(0, 1))
    maxs = np.max(frames, axis=(0, 1))

    centre = (mins + maxs) / 2
    half_range = np.max(maxs - mins) / 2 * 1.1

    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection="3d")

    scatter = ax.scatter(
        frames[0, :, 0],
        frames[0, :, 1],
        frames[0, :, 2],
        s=6
    )

    ax.set_xlim(
        centre[0] - half_range,
        centre[0] + half_range
    )
    ax.set_ylim(
        centre[1] - half_range,
        centre[1] + half_range
    )
    ax.set_zlim(
        centre[2] - half_range,
        centre[2] + half_range
    )

    ax.set_box_aspect((1, 1, 1))

    def update(frame):

        current = frames[frame]

        scatter._offsets3d = (
            current[:, 0],
            current[:, 1],
            current[:, 2]
        )

        ax.set_title(
            f"t = {frame * frame_skip * dt:.2f} s"
        )

        return scatter,

    animation = FuncAnimation(
        fig,
        update,
        frames=len(frames),
        interval=50
    )

    # save animation if a path is given
    if save_path is not None:
        animation.save(
            save_path,
            writer="pillow",
            fps=20
        )

    plt.close(fig)

    return HTML(animation.to_jshtml())
```

Whilst these outputs are functional, they lack much of the distinct behaviour of a starling flock. As mentioned in the introduction, starling flocks tend not to be in a spherical or regular pattern, rather an elliptical shape which we can easily improve on. Furthermore, given our scaling, it's clear that our birds relax instantly to a speed far slower than that which they would actually move at. Hence we can add a speed enforcing term, which allows for some flucturation centered around the expected speed. 

We now redefine our functions accordingly.

```python
import numpy as np

def calculate_acceleration_speed_enforced(
    position,
    velocity,
    alpha,
    beta,
    gamma,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2
):
    """
    positions, velocity Nx3 vectors
    """

    # compute neighbourhoods
    differences = position[:, None, :] - position[None, :, :] # differences[i,j] returns vector of differences

    distances = np.linalg.norm(differences, axis=2) # NxN matrix of absolute diffs
    np.fill_diagonal(distances, np.inf) # exclude themselves

    neighbour_indices = np.argsort(distances, axis=1)[:, :K]

    neighbour_positions = position[neighbour_indices] # NxKx3 of all neighbour positions
    neighbour_velocities = velocity[neighbour_indices] # NxKx3 of all neighbour velocities

    alignment = np.mean(neighbour_velocities, axis=1) - velocity
    cohesion = np.mean(neighbour_positions, axis=1) - position

    bird_indices = np.arange(len(position))[:, None]

    neighbour_differences = differences[bird_indices, neighbour_indices]
    neighbour_distances = distances[bird_indices, neighbour_indices]

    separation = np.sum(
        neighbour_differences / (neighbour_distances[:, :, None]**2 + epsilon**2),
        axis=1
    )

    # PREFERRED SPEED
    speed = np.linalg.norm(velocity, axis=1, keepdims=True)

    speed_control = (
        speed_strength
        * (preferred_speed - speed)
        * velocity
        / np.maximum(speed, 1e-12)
    )

    acceleration = (
        alpha * separation
        + beta * alignment
        + gamma * cohesion
        + speed_control
    )

    return acceleration
```

```python
def simulate_murmuration_speed_enforced(
    T=10,
    dt=0.02,
    N=500,
    alpha=0.2,
    beta=1.0,
    gamma=1.0,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2,
    seed=1,
    min_spacing=0.5,
):

    rng = np.random.default_rng(seed)

    # ELLIPTICAL FLOCK randomly generated. checks if in the ellipsis, if not regenerate. 
    semi_axes = np.array([6.0, 3.0, 1.2])

    positions = []

    while len(positions) < N:

        candidate = rng.uniform(
            -semi_axes,
            semi_axes
        )

        # candidate must be inside ellipsoid
        inside = np.sum((candidate / semi_axes)**2) <= 1

        if not inside:
            continue

        # candidate must not be too close to an existing bird
        if len(positions) > 0:
            existing = np.array(positions)

            distances = np.linalg.norm(
                existing - candidate,
                axis=1
            )

            if np.min(distances) < min_spacing:
                continue

        positions.append(candidate)

    position = np.array(positions)

    # initial velocities
    velocity = np.zeros_like(position)
    velocity[:, 0] = preferred_speed

    velocity += rng.uniform(
        -0.5,
        0.5,
        size=velocity.shape
    )

    # run the loop

    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    for step in range(n_steps):

        acceleration = calculate_acceleration_speed_enforced(
            position,
            velocity,
            alpha,
            beta,
            gamma,
            K,
            preferred_speed,
            speed_strength
        )

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```

We see that they start to fragment and lose cohesion at low betas, but at high betas we get regular "gas like" behaviour. We can check if it ever settles to equilibrium via a metric:

$$
\bar{\mathbf{x}}(t)
=
\frac{1}{N}
\sum_{i=1}^{N}
\mathbf{x}_i(t)
$$

$$
R(t)
=
\sqrt{
\frac{1}{N}
\sum_{i=1}^{N}
\left\|
\mathbf{x}_i(t)-\bar{\mathbf{x}}(t)
\right\|^2
}
$$

```python
def flock_radius(positions):

    centres = np.mean(
        positions,
        axis=1,
        keepdims=True
    )

    centred = positions - centres

    radius = np.sqrt(
        np.mean(
            np.sum(centred**2, axis=2),
            axis=1
        )
    )

    return radius
```

```python
plt.figure(figsize=(8, 5))

for beta, positions in positions_by_beta.items():

    radius = flock_radius(positions)

    times = np.arange(len(radius)) * 0.02

    plt.plot(
        times,
        radius,
        label=f"beta = {beta}"
    )

plt.xlabel("Time (s)")
plt.ylabel("Flock radius (m)")
plt.legend()
plt.show()
```

We see an interesting stability transition between 0.6 and 0.7, so for brevity we choose 0.65 as beta going forward. We can now attempt to simulate the turning of edge-situated birds in order to try and emulate the behaviour of the murmuration. We repatch our simulation functino to include such behaviour

We now introduce a simple local turning perturbation.

At a chosen time, we select a small proportion of birds from one elongated edge of the flock. These birds are then given an additional upward acceleration for a short period of time, after which the perturbation is removed.

This allows us to test whether a local change in direction can propagate through the flock using the existing interaction rules, or whether it is simply damped out or causes the flock to fragment.

```python
def simulate_murmuration_turn(
    T=30,
    dt=0.02,
    N=500,
    alpha=0.2,
    beta=0.65,
    gamma=1.0,
    K=6,
    preferred_speed=10.0,
    speed_strength=1.0,
    epsilon=0.2,
    seed=1,
    min_spacing=0.5,
    turn_start=5.0,
    turn_duration=0.5,
    turn_strength=5.0,
    turn_fraction=0.1
):

    rng = np.random.default_rng(seed)

    # ELLIPTICAL FLOCK randomly generated
    semi_axes = np.array([6.0, 3.0, 1.2])

    positions = []

    while len(positions) < N:

        candidate = rng.uniform(
            -semi_axes,
            semi_axes
        )

        inside = np.sum((candidate / semi_axes)**2) <= 1

        if not inside:
            continue

        if len(positions) > 0:

            existing = np.array(positions)

            distances = np.linalg.norm(
                existing - candidate,
                axis=1
            )

            if np.min(distances) < min_spacing:
                continue

        positions.append(candidate)

    position = np.array(positions)

    # initial velocities
    velocity = np.zeros_like(position)
    velocity[:, 0] = preferred_speed

    velocity += rng.uniform(
        -0.5,
        0.5,
        size=velocity.shape
    )

    # run the loop
    n_steps = int(T / dt)

    positions_over_time = [position.copy()]

    turning_birds = None

    for step in range(n_steps):

        time = step * dt

        acceleration = calculate_acceleration(
            position,
            velocity,
            alpha,
            beta,
            gamma,
            K,
            preferred_speed,
            speed_strength,
            epsilon
        )

        # choose birds at one elongated tip when the turn begins
        if time >= turn_start and turning_birds is None:

            threshold = np.quantile(
                position[:, 0],
                1 - turn_fraction
            )

            turning_birds = np.where(
                position[:, 0] >= threshold
            )[0]

        # apply temporary upward acceleration
        if (
            turning_birds is not None
            and time < turn_start + turn_duration
        ):

            acceleration[turning_birds, 2] += turn_strength

        velocity = velocity + acceleration * dt

        position = position + velocity * dt

        positions_over_time.append(position.copy())

    return np.array(positions_over_time)
```
